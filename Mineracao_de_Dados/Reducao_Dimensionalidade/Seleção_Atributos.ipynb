#### **Maldição da Dimensionalidade**

A maldição da dimensionalidade refere-se às severas dificuldades que surgem quando se trabalha com espaços que possuem muitas dimensões. À medida que a dimensionalidade ($D$) é aumentada, o número de regiões em uma grade regular cresce exponencialmente, tornando-se impraticável preencher esses espaços com dados suficientes para análises estatísticas. Geometricamente, as intuições desenvolvidas em três dimensões falham em espaços dimensionais superiores; por exemplo, em altas dimensões, a maior parte do volume de uma esfera concentra-se em uma casca fina perto de sua superfície e a massa de uma distribuição Gaussiana concentra-se em um anel fino a uma certa distância da origem.

#### **Redução de Dimensionalidade**

A redução de dimensionalidade é o processo de alterar a representação de dados complexos, originalmente descritos por um grande número de variáveis, para uma descrição de baixa dimensão usando apenas um pequeno número de parâmetros livres. Essa técnica pode ser implementada por meio de duas abordagens principais:

- **Seleção de atributos:** Identifica e seleciona o menor subconjunto de variáveis originais que melhor representam as características essenciais do problema, descartando as irrelevantes ou redundantes;
- **Extração de atributos:** Realiza transformações matemáticas ou combinações das variáveis originais para criar novas representações (como no método PCA), mantendo o máximo de informação possível em um número menor de dimensões.

A importância da redução de dimensionalidade reside em diversos fatores práticos e teóricos:

- **Eficiência Computacional e de Armazenamento:** Com a geração exponencial de dados (*Big Data*), a redução diminui a necessidade de espaço em disco e acelera drasticamente o tempo de processamento dos algoritmos;
- **Melhoria de Desempenho:** Ao eliminar redundâncias e ruídos ("informações irrelevantes"), as técnicas ajudam a evitar o sobreajuste (*overfitting*), permitindo que modelos de aprendizado de máquina alcancem maior precisão e generalização;
- **Visualização de Dados:** Projetar dados de alta dimensão em espaços de duas ou três dimensões permite que humanos visualizem estruturas, agrupamentos e padrões ocultos de forma intuitiva;
- **Estrutura Intrínseca:** Dados reais frequentemente residem em variedades não lineares (*manifolds*) de baixa dimensão embutidas no espaço de alta dimensão; a redução permite explorar essa estrutura efetiva inferior.

#### **Seleção de Atributos (Feature Selection)**

Esta é a abordagem mais intuitiva de redução, pois não transforma os dados originais, apenas escolhe o subconjunto mais relevante:

- Como funciona: Identifica e remove atributos irrelevantes ou redundantes que não contribuem para a definição da classe ou padrão. Pode ser feita via Filtros (estatísticas intrínsecas dos dados), *Wrappers* (usa algoritmos de predição para avaliar subconjuntos) ou Embutidos (seleção ocorre durante o treino do modelo); por exemplo: no diagnóstico de doenças genéticas, a técnica é usada para selecionar apenas alguns genes específicos entre milhares presentes em um microarranjo de DNA, facilitando a identificação de padrões de câncer.

A aplicação da técnica segue um ciclo estruturado em quatro passos fundamentais:

1. **Geração do Subconjunto**: utiliza-se uma estratégia de busca para controlar a criação de novos subconjuntos de candidatos. As estratégias podem variar desde buscas gulosas, como a seleção para frente (*forward*), que começa vazia e adiciona atributos um a um, até a eliminação para trás, que começa com todos e remove o pior. Ou métodos mais complexos como Algoritmos Genéticos.
2. **Avaliação do Subconjunto**: Cada subconjunto gerado é testado por uma medida de avaliação para determinar sua qualidade. Essa etapa define a abordagem utilizada:

    - Filtro: Avalia os atributos com base apenas em características intrínsecas dos dados (como correlação ou ganho de informação), sem envolver algoritmos de aprendizado de máquina.
    - *Wrapper*: Utiliza um algoritmo de indução (classificador) como uma "caixa-preta" para avaliar o desempenho do subconjunto, buscando minimizar a taxa de erro.
    - Embutido: A seleção ocorre automaticamente durante a fase de construção ou treinamento do próprio modelo de classificação (como em árvores de decisão).

3. **Critério de Parada**: O ciclo de geração e avaliação continua até que um critério pré-definido seja satisfeito, como atingir um número específico de atributos ou não haver mais melhora na performance.

4. **Validação dos Resultados**: Uma vez interrompida a busca, o subconjunto final é validado em um conjunto de testes independente para garantir sua capacidade de generalização.

Essa técnica mitiga a maldição da dimensionalidade, reduz o custo computacional e evita o sobreajuste (*overfitting*), permitindo que o modelo aprenda padrões reais em vez de ruídos nos dados.

```python
import pandas as pd 
import numpy as np
```

```python
dados = pd.read_csv('Conjunto_Dados/Human_Activity_Recognition_with_Smartphones.csv')
```

```python
dados.head(10)
```

```python
dados.shape
```

```python
dados['Activity'].unique()

mapa = {
    'STANDING': 0,
    'SITTING': 1,
    'WALKING_DOWNSTAIRS': 2,
    'LAYING': 3,
    'WALKING': 4,
    'WALKING_UPSTAIRS': 5
}

dados['Activity'] = dados['Activity'].map(mapa)
```

```python
from sklearn.feature_selection import VarianceThreshold

# VarianceThreshold é um método de seleção de atributos que remove todas as características que têm uma variância abaixo de um determinado limiar.
selector = VarianceThreshold(threshold=0.01)

dados_r = selector.fit_transform(dados)
dados_reduzido = pd.DataFrame(dados_r, columns=dados.columns[selector.get_support()])

print("Antes:", dados.shape)
print("Depois:", dados_reduzido.shape)
```

```python
# Matriz de Correlação de Pearson.
matriz_correlacao = dados_reduzido.corr().abs()

# Matriz triangular superior para evitar a duplicação de correlações, pois a matriz de correlação é simétrica.
superior = matriz_correlacao.where(np.triu(np.ones(matriz_correlacao.shape), k=1).astype(bool))

# Colunas com alta correlação, ou seja, aquelas que têm uma correlação maior que 0.95.
dropar = [col for col in superior.columns if any(superior[col] > 0.95)]

# Removendo as colunas altamente correlacionadas.
dados_reduzido = dados_reduzido.drop(columns=dropar)

print("Depois:", dados_reduzido.shape)
```

Agora que já eliminamos algumas *features* básicas super correlacionadas ou quase constantes, vamos eliminar mais algumas features por meio de um *filter* (filtro) ou *wrapper* (envelopador).

##### a) **Filtro**
Avalia os atributos com base apenas em características intrínsecas dos dados (como correlação ou ganho de informação), sem envolver algoritmos de aprendizado de máquina (SelectKBest, correlação, mutual information, ANOVA).

```python
# Dividindo os dados em conjuntos de treinamento e teste.
from sklearn.model_selection import train_test_split, GridSearchCV
# Pipeline para seleção de atributos e classificação.
from sklearn.pipeline import Pipeline
# Seleção de atributos usando SelectKBest e classificação usando RandomForestClassifier.
from sklearn.feature_selection import SelectKBest, f_classif
# Classificador Random Forest.
from sklearn.ensemble import RandomForestClassifier

# Dividindo os dados em conjuntos de treinamento e teste.
X_train, X_test, y_train, y_test = train_test_split(
    dados_reduzido.drop(columns=["Activity"]),
    dados_reduzido["Activity"],
    test_size=0.2,
    random_state=42,
    stratify=dados_reduzido["Activity"]
)

# Pipeline para seleção de atributos e classificação.
pipe = Pipeline([
    ("selecao_coluna", SelectKBest(score_func=f_classif)),
    ("classificador", RandomForestClassifier(random_state=42))
])

# GridSearchCV para encontrar o melhor número de atributos a serem selecionados.
param_grid = {
    "selecao_coluna__k": [20, 50, 100, 200, 250]
}

# GridSearchCV para encontrar o melhor número de atributos a serem selecionados.
grid = GridSearchCV(
    pipe,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

# Treinando o modelo com os dados de treinamento.
grid.fit(X_train, y_train)

# Resultados do GridSearchCV.
print("Melhor k:", grid.best_params_["selecao_coluna__k"])
print("Acurácia CV:", grid.best_score_)

# Avaliando o modelo no conjunto de teste.
best_model = grid.best_estimator_
test_score = best_model.score(X_test, y_test)

print("Acurácia no teste:", test_score)
```

##### b) **Wrapper**
Utiliza um algoritmo de indução (classificador) como uma "caixa-preta" para avaliar o desempenho do subconjunto, buscando minimizar a taxa de erro (Forward Selection, RFE, Sequential Feature Selection). Abaixo está um exemplo de aplicação para o Conjunto de Dados abordado nesse laboratório, entretanto, para esse caso o treino fica excessivamente longo por conta do número de colunas ($500$+), o interessante seria rodar essa técnica com até no máximo em dados com cerca de $100$ colunas.

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.svm import SVC

# classificador SVC com kernel linear.
svc = SVC(kernel="linear")

# Seletor de atributos sequencial para selecionar as 50 melhores características.
sfs = SequentialFeatureSelector(
    estimator=svc,
    n_features_to_select=50,
    direction="forward",
    cv=5,
    n_jobs=-1
)

# Treinando o modelo com os dados de treinamento.
sfs.fit(X_train, y_train)

# Transformando os dados de treinamento e teste para conter apenas as 50 características selecionadas.
X_train_reduzido = sfs.transform(X_train)
X_test_reduzido = sfs.transform(X_test)
```
