#### **Maldição da Dimensionalidade**

A maldição da dimensionalidade refere-se às severas dificuldades que surgem quando se trabalha com espaços que possuem muitas dimensões. À medida que a dimensionalidade ($D$) é aumentada, o número de regiões em uma grade regular cresce exponencialmente, tornando-se impraticável preencher esses espaços com dados suficientes para análises estatísticas. Geometricamente, as intuições desenvolvidas em três dimensões falham em espaços dimensionais superiores; por exemplo, em altas dimensões, a maior parte do volume de uma esfera concentra-se em uma casca fina perto de sua superfície e a massa de uma distribuição Gaussiana concentra-se em um anel fino a uma certa distância da origem.

#### **Redução de Dimensionalidade**

A redução de dimensionalidade é o processo de alterar a representação de dados complexos, originalmente descritos por um grande número de variáveis, para uma descrição de baixa dimensão usando apenas um pequeno número de parâmetros livres. Essa técnica pode ser implementada por meio de duas abordagens principais:

- **Seleção de atributos:** Identifica e seleciona o menor subconjunto de variáveis originais que melhor representam as características essenciais do problema, descartando as irrelevantes ou redundantes;
- **Extração de atributos:** Realiza transformações matemáticas ou combinações das variáveis originais para criar novas representações (como no método PCA), mantendo o máximo de informação possível em um número menor de dimensões.

A importância da redução de dimensionalidade reside em diversos fatores práticos e teóricos:

- **Eficiência Computacional e de Armazenamento:** Com a geração exponencial de dados (*Big Data*), a redução diminui a necessidade de espaço em disco e acelera drasticamente o tempo de processamento dos algoritmos;
- **Melhoria de Desempenho:** Ao eliminar redundâncias e ruídos ("informações irrelevantes"), as técnicas ajudam a evitar o sobreajuste (*overfitting*), permitindo que modelos de aprendizado de máquina alcancem maior precisão e generalização;
- **Visualização de Dados:** Projetar dados de alta dimensão em espaços de duas ou três dimensões permite que humanos visualizem estruturas, agrupamentos e padrões ocultos de forma intuitiva;
- **Estrutura Intrínseca:** Dados reais frequentemente residem em variedades não lineares (*manifolds*) de baixa dimensão embutidas no espaço de alta dimensão; a redução permite explorar essa estrutura efetiva inferior.

#### **Kernel PCA**

O Kernel Principal Component Analysis (Kernel PCA) é uma extensão não linear da Análise de Componentes Principais (PCA) que utiliza a técnica de substituição de kernel (ou "truque do kernel") para capturar estruturas complexas em conjuntos de dados que não são linearmente separáveis no espaço original. Formalmente, o Kernel PCA realiza uma transformação não linear $\phi(x)$ dos dados originais para um espaço de características (*feature space*) de alta dimensionalidade (podendo ser infinita). Nesse novo espaço, o algoritmo executa um PCA linear padrão, o que equivale implicitamente a encontrar componentes principais não lineares no espaço de dados original. A técnica baseia-se no fato de que o PCA pode ser formulado inteiramente em termos de produtos escalares entre vetores de dados. O Kernel PCA substitui esses produtos escalares $x^Tx′$ por uma função de kernel $k(x,x′) = \phi(x)^T\phi(x′)$, permitindo que o mapeamento para o espaço de alta dimensão ocorra sem a necessidade de calcular explicitamente as coordenadas de $\phi(x)$.

A aplicação do Kernel PCA segue um processo matemático baseado na decomposição espectral de uma matriz de produtos internos no espaço de características implícito:

1. **Escolha do Kernel**: Seleciona-se uma função de kernel apropriada para os dados, como o Kernel Gaussiano (RBF), Polinomial ou Sigmoidal. O kernel define implicitamente o espaço de alta dimensão onde o PCA será realizado.

2. **Cálculo da Matriz de Gram (K)**: Constrói-se uma matriz $N \times N$, onde cada elemento $K_{nm}$ é dado por:
$$
K_{nm} = k(x_n, x_m)
$$
representando um produto interno no espaço de características induzido pelo kernel.

3. **Centralização da Matriz de Gram**: Como o PCA assume dados centralizados, aplica-se centralização na matriz de kernel:
$$
K' = K - \mathbf{1}K - K\mathbf{1} + \mathbf{1}K\mathbf{1}
$$
onde $\mathbf{1}$ é a matriz de médias $ \frac{1}{N} $.

4. **Decomposição em Autovalores**: Resolve-se o problema de autovalores da matriz centralizada $K'$:
$$
K' a_i = \lambda_i a_i
$$
obtendo autovetores $a_i$ e autovalores $\lambda_i$.

5. **Projeção dos Dados**: As projeções de um ponto $x$ no i-ésimo componente principal são dadas por:
$$
y_i(x) = \frac{1}{\sqrt{\lambda_i}} \sum_{n=1}^{N} a_{in} \, k(x, x_n)
$$

#### Aplicação do Kernel PCA

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import KernelPCA
from sklearn.model_selection import train_test_split

X_train = dados.drop('Activity', axis=1).values
y_train = dados['Activity'].values

X_train, X_test, y_train, y_test = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42, stratify=y_train
)

# Normalização dos dados.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Aplicação do Kernel PCA.
kpca = KernelPCA(
    n_components=50,     # Número de componentes principais a serem mantidos.
    kernel="rbf",        # Tipo de kernel a ser usado (RBF é uma escolha comum para dados não lineares).
    gamma=0.01,          # Parâmetro do kernel RBF (ajuste conforme necessário).
    fit_inverse_transform=False
)

X_train_kpca = kpca.fit_transform(X_train_scaled)
X_test_kpca = kpca.transform(X_test_scaled)

print("Dimensão original:", X_train.shape[1])
print("Dimensão reduzida:", X_train_kpca.shape[1])
```
