#### **Maldição da Dimensionalidade**

A maldição da dimensionalidade refere-se às severas dificuldades que surgem quando se trabalha com espaços que possuem muitas dimensões. À medida que a dimensionalidade ($D$) é aumentada, o número de regiões em uma grade regular cresce exponencialmente, tornando-se impraticável preencher esses espaços com dados suficientes para análises estatísticas. Geometricamente, as intuições desenvolvidas em três dimensões falham em espaços dimensionais superiores; por exemplo, em altas dimensões, a maior parte do volume de uma esfera concentra-se em uma casca fina perto de sua superfície e a massa de uma distribuição Gaussiana concentra-se em um anel fino a uma certa distância da origem.

#### **Redução de Dimensionalidade**

A redução de dimensionalidade é o processo de alterar a representação de dados complexos, originalmente descritos por um grande número de variáveis, para uma descrição de baixa dimensão usando apenas um pequeno número de parâmetros livres. Essa técnica pode ser implementada por meio de duas abordagens principais:

- **Seleção de atributos:** Identifica e seleciona o menor subconjunto de variáveis originais que melhor representam as características essenciais do problema, descartando as irrelevantes ou redundantes;
- **Extração de atributos:** Realiza transformações matemáticas ou combinações das variáveis originais para criar novas representações (como no método PCA), mantendo o máximo de informação possível em um número menor de dimensões.

A importância da redução de dimensionalidade reside em diversos fatores práticos e teóricos:

- **Eficiência Computacional e de Armazenamento:** Com a geração exponencial de dados (*Big Data*), a redução diminui a necessidade de espaço em disco e acelera drasticamente o tempo de processamento dos algoritmos;
- **Melhoria de Desempenho:** Ao eliminar redundâncias e ruídos ("informações irrelevantes"), as técnicas ajudam a evitar o sobreajuste (*overfitting*), permitindo que modelos de aprendizado de máquina alcancem maior precisão e generalização;
- **Visualização de Dados:** Projetar dados de alta dimensão em espaços de duas ou três dimensões permite que humanos visualizem estruturas, agrupamentos e padrões ocultos de forma intuitiva;
- **Estrutura Intrínseca:** Dados reais frequentemente residem em variedades não lineares (*manifolds*) de baixa dimensão embutidas no espaço de alta dimensão; a redução permite explorar essa estrutura efetiva inferior.

#### **AutoEncoders**

Os AutoEncoders (também chamados de redes autoassociativas) são uma técnica de aprendizado não supervisionado baseada em redes neurais multicamadas, projetada para converter dados de alta dimensionalidade em representações compactas de baixa dimensão, chamadas de códigos. Formalmente, um autoencoder é definido como um modelo que busca aprender uma mapeamento autoassociativo, onde a rede é treinada para mapear vetores de entrada nela mesma. A arquitetura consiste em dois componentes fundamentais:

- Codificador (*Encoder*): Uma função (ou rede) que transforma os dados de alta dimensionalidade $x$ em um código latente $z$ de dimensão inferior;
- Decodificador (*Decoder*): Uma função que tenta recuperar ou reconstruir os dados originais $x$ a partir do código $z$.

O objetivo matemático é determinar os parâmetros (pesos e vieses) da rede minimizando uma função de erro de reconstrução, que captura a discrepância entre o vetor original e sua reconstrução. Frequentemente, utiliza-se a soma dos erros quadráticos para essa finalidade. Quando as camadas ocultas utilizam ativações lineares, o autoencoder é equivalente à Análise de Componentes Principais (PCA); no entanto, com camadas não lineares, ele atua como uma generalização não linear do PCA, capaz de revelar estruturas complexas nos dados.

A aplicação de Autoencoders para redução de dimensionalidade segue geralmente estes passos estruturais:

- **Definição da Arquitetura**: Projeta-se uma rede neural com um gargalo (camada latente de menor dimensão), forçando a compressão da informação;

- **Inicialização**: Em práticas modernas, utiliza-se inicialização padrão (ex: He ou Glorot). O pré-treinamento camada a camada (ex: RBMs) é historicamente relevante, mas raramente utilizado hoje devido à eficácia do treinamento direto com backpropagation;

- **Treinamento (Reconstrução)**: O modelo é treinado para minimizar o erro de reconstrução:
$$X \approx \hat{X}$$
usando *backpropagation* e otimizadores como Adam (Adaptive Moment Estimation);

- **Ajuste Fino (Fine-tuning)**: O treinamento já ocorre de forma conjunta em todas as camadas, não sendo necessário separar etapas como no pré-treinamento clássico;

- **Variante Variacional (VAE)**: Em contextos probabilísticos, o encoder aprende uma distribuição no espaço latente $(\mu, \sigma)$, e utiliza-se o truque da reparametrização. A função de perda combina:
    - Erro de reconstrução;
    - Divergência KL (regularização do espaço latente).

Uma vez treinado, o **encoder** é utilizado isoladamente para mapear dados de alta dimensão em um espaço latente reduzido.

---

#### Pipeline de Autoencoder (prático)

1. Pré-processamento:
   - Normalização (StandardScaler).

2. Definição da arquitetura:
   - Encoder (redução);
   - Decoder (reconstrução).

3. Compilação:
   - Perda: MSE (erro quadrático médio);
   - Otimizador: Adam.

4. Treinamento:
   - X -> X (entrada = saída)

5. Extração:
   - Usar apenas o encoder;

Pode ser usado para treinamento de modelo preditivo e de agrupamento, bem como para visualização em alguns casos.

### **AutoEncoder**

#### 1. Normalização

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Separar as features (X) e o alvo (y).
X = dados.drop('Activity', axis=1)
y = dados['Activity']

# Padronizar as features.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Separar os dados em conjuntos de treino e teste, mantendo a proporção das classes.
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)
```

#### 2. Construção do AutoEncoder

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

# Definir a dimensão de entrada e a dimensão de codificação (redução).
dimensao_entrada = X_train.shape[1]
dimensao_codificacao = 32  # Isso pode ser ajustado conforme necessário, é o gargalo da redução.

# Entrada: a mesma dimensão dos dados originais.
camada_entrada = Input(shape=(dimensao_entrada,))

# Codificador (Encoder): camadas densas que reduzem a dimensionalidade, com funções de ativação ReLU para introduzir não linearidade.
codificado = Dense(128, activation='relu')(camada_entrada) 
codificado = Dense(128, activation='relu')(camada_entrada) 
codificado = Dense(64, activation='relu')(codificado)
codificado = Dense(dimensao_codificacao, activation='relu')(codificado)

# Decodificador (Decoder): camadas densas que tentam reconstruir os dados originais a partir da representação 
# codificada, com função de ativação linear na última camada para permitir a reconstrução dos valores originais.
decodificado = Dense(64, activation='relu')(codificado)
decodificado = Dense(128, activation='relu')(decodificado)
decodificado = Dense(dimensao_entrada, activation='linear')(decodificado)

# Modelos: o autoencoder completo para treinamento e o codificador separado para obter as representações reduzidas.
autoencoder = Model(camada_entrada, decodificado)
codificador = Model(camada_entrada, codificado)
```

#### 3. Compilar

```python
# Compilar o autoencoder: usamos o otimizador Adam, que é eficiente para muitos problemas de otimização, e a função de perda MSE (Mean Squared Error).
autoencoder.compile(
    optimizer='adam',
    loss='mse'
)
```

#### 4. Treinamento

```python
# Treinar o autoencoder: o modelo é treinado para minimizar a diferença entre os dados de entrada e os dados reconstruídos, 
# usando um conjunto de validação para monitorar o desempenho durante o treinamento.

autoencoder.fit(
    X_train, X_train,
    epochs=50,
    batch_size=256,
    validation_split=0.2,
    verbose=1
)
```

#### 5. Redução de dimensionalidade

```python
X_train_encoded = codificador.predict(X_train)
X_test_encoded = codificador.predict(X_test)

print(X_train.shape)        
print(X_train_encoded.shape) 
```

```python
dados_reduzidos = pd.DataFrame(X_train_encoded)
dados_reduzidos['Activity'] = y_train.values
dados_reduzidos.head(10)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# =========================
# 1. MODELO NO ESPAÇO ORIGINAL
# =========================
clf1 = RandomForestClassifier(n_estimators=100, random_state=42)

clf1.fit(X_train, y_train)
y_pred1 = clf1.predict(X_test)

acc_original = accuracy_score(y_test, y_pred1)


# =========================
# 2. MODELO NO ESPAÇO COMPRIMIDO
# =========================
clf2 = RandomForestClassifier(n_estimators=100, random_state=42)

clf2.fit(X_train_encoded, y_train)
y_pred2 = clf2.predict(X_test_encoded)

acc_encoded = accuracy_score(y_test, y_pred2)


# =========================
# 3. RESULTADOS
# =========================
print("Accuracy original:", acc_original)
print("Accuracy encoded:", acc_encoded)

```
