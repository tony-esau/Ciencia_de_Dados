#### **Maldição da Dimensionalidade**

A maldição da dimensionalidade refere-se às severas dificuldades que surgem quando se trabalha com espaços que possuem muitas dimensões. À medida que a dimensionalidade ($D$) é aumentada, o número de regiões em uma grade regular cresce exponencialmente, tornando-se impraticável preencher esses espaços com dados suficientes para análises estatísticas. Geometricamente, as intuições desenvolvidas em três dimensões falham em espaços dimensionais superiores; por exemplo, em altas dimensões, a maior parte do volume de uma esfera concentra-se em uma casca fina perto de sua superfície e a massa de uma distribuição Gaussiana concentra-se em um anel fino a uma certa distância da origem.

#### **Redução de Dimensionalidade**

A redução de dimensionalidade é o processo de alterar a representação de dados complexos, originalmente descritos por um grande número de variáveis, para uma descrição de baixa dimensão usando apenas um pequeno número de parâmetros livres. Essa técnica pode ser implementada por meio de duas abordagens principais:

- **Seleção de atributos:** Identifica e seleciona o menor subconjunto de variáveis originais que melhor representam as características essenciais do problema, descartando as irrelevantes ou redundantes;
- **Extração de atributos:** Realiza transformações matemáticas ou combinações das variáveis originais para criar novas representações (como no método PCA), mantendo o máximo de informação possível em um número menor de dimensões.

A importância da redução de dimensionalidade reside em diversos fatores práticos e teóricos:

- **Eficiência Computacional e de Armazenamento:** Com a geração exponencial de dados (*Big Data*), a redução diminui a necessidade de espaço em disco e acelera drasticamente o tempo de processamento dos algoritmos;
- **Melhoria de Desempenho:** Ao eliminar redundâncias e ruídos ("informações irrelevantes"), as técnicas ajudam a evitar o sobreajuste (*overfitting*), permitindo que modelos de aprendizado de máquina alcancem maior precisão e generalização;
- **Visualização de Dados:** Projetar dados de alta dimensão em espaços de duas ou três dimensões permite que humanos visualizem estruturas, agrupamentos e padrões ocultos de forma intuitiva;
- **Estrutura Intrínseca:** Dados reais frequentemente residem em variedades não lineares (*manifolds*) de baixa dimensão embutidas no espaço de alta dimensão; a redução permite explorar essa estrutura efetiva inferior.

#### **Incorporação Linear Local - Locally Linear Embedding (LLE)**

O Locally Linear Embedding (LLE) é uma técnica de redução de dimensionalidade não linear que visa preservar as propriedades geométricas locais da vizinhança dos dados. O algoritmo baseia-se na premissa de que cada ponto de dado e seus vizinhos residem em uma região que pode ser considerada aproximadamente linear da variedade (*manifold*). Formalmente, o LLE é equivalente a encontrar as autofunções do quadrado do Laplaciano do grafo normalizado, o que coincide com a diagonalização do próprio Laplaciano do grafo.

Aplicação:

- Descoberta da vizinhança: Para cada ponto de dado $_xi$​, identificam-se seus $k$ vizinhos mais próximos;
- Construção da matriz de aproximação: Calculam-se os pesos $W_{ij}$​ que melhor reconstroem cada ponto $x_i$​ a partir de seus vizinhos, minimizando o erro de reconstrução quadrático sob a restrição de que a soma dos pesos para cada ponto seja igual a $1$. Esses pesos são invariantes a rotações, translações e escalas;
- Cálculo do embutimento (embedding): As coordenadas de baixa dimensão $y_i$​ são obtidas extraindo-se os autovetores correspondentes aos menores autovalores da matriz $(I−W)^T(I−W)$, que pode ser interpretada como o Laplaciano iterado $L^2$.

#### **Aplicação do Locally Linear Embedding**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Separação entre variáveis independentes (X) e variável dependente (y).
X = dados.drop('Activity', axis=1)
y = dados['Activity']

# Escalonamento dos dados: média 0 e desvio padrão 1.
escalonador = StandardScaler() 
X_escalonado = escalonador.fit_transform(X)

# Divisão dos dados em conjuntos de treinamento e teste, mantendo a proporção das classes.
X_train, X_test, y_train, y_test = train_test_split(
    X_escalonado, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
from sklearn.manifold import LocallyLinearEmbedding

lle = LocallyLinearEmbedding(
    n_neighbors=10,     # Número de vizinhos (hiperparâmetro crítico).
    n_components=2,     # Dimensão Final: 2 para visualização.
    method='standard',
    n_jobs=-1
)

X_train_lle = lle.fit_transform(X_train)
X_test_lle = lle.transform(X_test)
dados_reduzidos = pd.DataFrame(X_train_lle, columns=['Dim1', 'Dim2'])
dados_reduzidos['Activity'] = y_train.values
```

```python
dados_reduzidos.head()
```

```python
lle = LocallyLinearEmbedding(
    n_neighbors=10,     # Número de vizinhos (Precisa ser ajustado conforme o dataset).
    n_components=20,     # Dimensão para treino: 5, 10, 20, 50, 100. (Teste para encontrar o melhor).
    method='standard',
    n_jobs=-1
)

X_train_lle = lle.fit_transform(X_train)
X_test_lle = lle.transform(X_test)
dados_reduzidos = pd.DataFrame(X_train_lle, columns=[f'Dim{i+1}' for i in range(X_train_lle.shape[1])])
dados_reduzidos['Activity'] = y_train.values
```

#### **Relatório: Locally Linear Embedding (LLE)**

##### **Visão geral**

O **LLE** é um método de redução de dimensionalidade não linear baseado na hipótese de que os dados estão distribuídos sobre um **manifold suave** de baixa dimensão embutido em um espaço de alta dimensão.

Ele preserva **relações locais** ao reconstruir cada ponto como combinação linear de seus vizinhos mais próximos.


##### **O que o LLE realiza:**

- Preserva **estrutura local** (vizinhança): os pontos são obtidos como combinação linear dos seus vizinhos;
- Não assume linearidade global;
- Produz um **embedding de baixa dimensão**;
- Baseia-se em:
  - reconstrução local (pesos \(W\));
  - problema espectral global.

##### **Quando usar?**

**Cenários ideais**
- Dados com **estrutura não linear (manifold)**;
- Quando PCA falha em capturar a geometria;
- Análise exploratória de dados complexos;
- Visualização de dados de alta dimensão;
- Dados onde **relações locais são mais importantes que globais**.

**Exemplos**
- Sensores;
- Imagens;
- Dados biomédicos;
- Dados com clusters curvos ou superfícies.

##### **Quando não usar?**

**Cenários problemáticos**

- Dados muito ruidosos;
- Alta dimensionalidade com pouco sinal;
- Dataset muito grande (custo computacional alto);
- Quando relações globais são importantes;
- Quando precisa de interpretabilidade.

##### **Vantagens**

- Captura **estruturas não lineares**;
- Preserva relações locais com alta fidelidade;
- Não depende de funções de kernel;
- Invariância a:
  - translação;
  - rotação;
  - escala uniforme;
- Boa visualização de *manifolds*;
- Revela estrutura escondida nos dados;
- Melhora separação em problemas não lineares;
- Reduz dimensionalidade sem perder geometria local;
- Pode melhorar desempenho de modelos.

##### **Desvantagens**

- Sensível ao parâmetro **n_neighbors**;
- Não escala bem para grandes datasets;
- Sensível a ruído;
- Não supervisionado;
- Não preserva estrutura global;
- Pode gerar *embeddings* instáveis;
- Não possui métrica clara de qualidade (como variância explicada).

##### **Hiperparâmetros críticos**

**n_neighbors**

- Define estrutura local;
- Muito pequeno: quebra o *manifold*;
- Muito grande: perde não linearidade.

**n_components**

- Dimensão final;
- Definido empiricamente.

##### O que se obtém ao fim da aplicação:

- Um novo conjunto:
$$X \in \mathbb{R}^{d} \rightarrow Z \in \mathbb{R}^{k}, \quad k \ll d$$

- Cada ponto:
  - representa posição no *manifold*;
  - preserva relações locais.


##### **Comparação com outros métodos**

| Método | Tipo | Preserva | Escalabilidade | Uso típico |
|--------|------|----------|---------------|-----------|
| PCA | Linear | Variância global | Alta | Redução geral |
| Kernel PCA | Não linear | Estrutura global implícita | Média | Dados não lineares |
| MDS | Não linear | Distâncias | Baixa | Visualização |
| **LLE** | Não linear | Estrutura local | Baixa | Manifolds |
