#### **Maldição da Dimensionalidade**

A maldição da dimensionalidade refere-se às severas dificuldades que surgem quando se trabalha com espaços que possuem muitas dimensões. À medida que a dimensionalidade ($D$) é aumentada, o número de regiões em uma grade regular cresce exponencialmente, tornando-se impraticável preencher esses espaços com dados suficientes para análises estatísticas. Geometricamente, as intuições desenvolvidas em três dimensões falham em espaços dimensionais superiores; por exemplo, em altas dimensões, a maior parte do volume de uma esfera concentra-se em uma casca fina perto de sua superfície e a massa de uma distribuição Gaussiana concentra-se em um anel fino a uma certa distância da origem.

#### **Redução de Dimensionalidade**

A redução de dimensionalidade é o processo de alterar a representação de dados complexos, originalmente descritos por um grande número de variáveis, para uma descrição de baixa dimensão usando apenas um pequeno número de parâmetros livres. Essa técnica pode ser implementada por meio de duas abordagens principais:

- **Seleção de atributos:** Identifica e seleciona o menor subconjunto de variáveis originais que melhor representam as características essenciais do problema, descartando as irrelevantes ou redundantes;
- **Extração de atributos:** Realiza transformações matemáticas ou combinações das variáveis originais para criar novas representações (como no método PCA), mantendo o máximo de informação possível em um número menor de dimensões.

A importância da redução de dimensionalidade reside em diversos fatores práticos e teóricos:

- **Eficiência Computacional e de Armazenamento:** Com a geração exponencial de dados (*Big Data*), a redução diminui a necessidade de espaço em disco e acelera drasticamente o tempo de processamento dos algoritmos;
- **Melhoria de Desempenho:** Ao eliminar redundâncias e ruídos ("informações irrelevantes"), as técnicas ajudam a evitar o sobreajuste (*overfitting*), permitindo que modelos de aprendizado de máquina alcancem maior precisão e generalização;
- **Visualização de Dados:** Projetar dados de alta dimensão em espaços de duas ou três dimensões permite que humanos visualizem estruturas, agrupamentos e padrões ocultos de forma intuitiva;
- **Estrutura Intrínseca:** Dados reais frequentemente residem em variedades não lineares (*manifolds*) de baixa dimensão embutidas no espaço de alta dimensão; a redução permite explorar essa estrutura efetiva inferior.

#### **t-Distributed Stochastic Neighbor Embedding (t-SNE)**

É uma técnica de redução de dimensionalidade não linear projetada especificamente para a visualização de conjuntos de dados de alta dimensão em espaços de duas ou três dimensões. Formalmente, o t-SNE visa encontrar uma representação de baixa dimensão que minimize a divergência de Kullback-Leibler (KL) entre duas distribuições de probabilidade: uma distribuição conjunta ($P$) que descreve as semelhanças entre pontos no espaço original e uma distribuição ($Q$) que descreve as semelhanças no espaço reduzido. O diferencial da técnica em relação ao seu antecessor (SNE) é o uso de uma distribuição $t$ de Student com um grau de liberdade (equivalente a uma distribuição de Cauchy) para calcular as afinidades no espaço de baixa dimensão. Essa distribuição possui "caudas pesadas", o que permite que pontos moderadamente distantes no espaço original sejam representados por distâncias muito maiores no mapa reduzido, resolvendo efetivamente o "problema do amontoamento" (crowding problem) e facilitando a formação de agrupamentos nítidos.

A aplicação do t-SNE segue um processo iterativo baseado em forças de atração e repulsão:

1. **Cálculo de Afinidades de Alta Dimensão**: Converte as distâncias euclidianas entre os pontos originais em probabilidades condicionais $p_{j∣i}$​ usando uma distribuição Gaussiana. O parâmetro fundamental aqui é a perplexidade, que atua como uma medida suave do número efetivo de vizinhos para cada ponto;
2. **Simetrização das Probabilidades**: Define uma probabilidade conjunta simétrica $p_{ij}$​ para garantir que cada ponto, mesmo que isolado (*outlier*), contribua significativamente para a função de custo;
3. **Inicialização**: As coordenadas iniciais no espaço de baixa dimensão (mapa) são geradas aleatoriamente, geralmente amostradas de uma distribuição normal com variância muito pequena próxima à origem;
4. **Otimização por Descida de Gradiente**: O algoritmo ajusta as posições dos pontos no mapa de forma iterativa para minimizar a divergência $KL$. Durante esse processo:
    - Pontos similares no espaço original exercem uma força de atração entre si no mapa;
    - Pontos dissimilar exercem uma força de repulsão, impedindo que fiquem muito próximos.
5. ***Early Exaggeration* (Exagero Inicial)**: Frequentemente, nas primeiras $50$ a $100$ iterações, as probabilidades $p_{ij}$​ são multiplicadas por um fator constante (geralmente $4$ ou $12$). Isso cria agrupamentos muito apertados e espaços vazios amplos, permitindo que os clusters se movam uns através dos outros para encontrar uma organização global melhor antes do refinamento local.

```python
import pandas as pd 
import numpy as np
```

```python
dados = pd.read_csv('Conjunto_Dados/Human_Activity_Recognition_with_Smartphones.csv')
```

```python
dados.head(10)
```

```python
dados.shape
```

```python
dados['Activity'].unique()

mapa = {
    'STANDING': 0,
    'SITTING': 1,
    'WALKING_DOWNSTAIRS': 2,
    'LAYING': 3,
    'WALKING': 4,
    'WALKING_UPSTAIRS': 5
}

dados['Activity'] = dados['Activity'].map(mapa)
```

---

#### **Aplicação do Laplacian Eigenmaps**

```python
# Separando o X e o y.

X = dados.drop('Activity', axis=1)
y = dados['Activity']
```

```python
from sklearn.preprocessing import StandardScaler

# Padronizando os dados para que tenham média 0 e desvio padrão 1.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(
    n_components=2,        # Dimensão de saída.
    perplexity=30,         # Perplexidade, que é um parâmetro relacionado ao número de vizinhos considerados.
    learning_rate='auto',
    max_iter=1000, # Número máximo de iterações para otimização.
    random_state=42,
    init='pca' # Método de inicialização, que pode ser 'random' ou 'pca'.
)

X_tsne = tsne.fit_transform(X_scaled)
dados_reduzidos = pd.DataFrame(X_tsne, columns=['Dimensão 1', 'Dimensão 2'])
dados_reduzidos['Activity'] = y
```

#### **t-SNE: Relatório Técnico de Uso e Aplicação**

##### **Objetivo prático**

O **t-SNE** é utilizado para:

- **Visualizar dados de alta dimensão**;
- **Identificar clusters e padrões ocultos**;
- **Análise exploratória (EDA)**.

> Não é uma técnica voltada para modelagem preditiva.

#### **Quando usar**

- Dados com **alta dimensionalidade** (ex: $>50$ features);
- Estruturas **não lineares**;
- Quando PCA não separa bem os dados;
- Interesse em:
  - visualizar clusters;
  - entender estrutura dos dados;

##### **Exemplos práticos**

- Imagens (embeddings);
- NLP (word embeddings);
- Bioinformática (gene expression).

##### **Quando não usar**

- Precisa treinar modelo diretamente com os dados reduzidos;
- Precisa aplicar em novos dados;
- Quer interpretar distâncias numericamente;
- Dataset com muitas amostras.
