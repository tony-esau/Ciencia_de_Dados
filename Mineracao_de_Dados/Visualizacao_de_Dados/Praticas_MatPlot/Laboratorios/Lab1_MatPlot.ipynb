##### **Normalização (em base 100)**
Quando se comparam duas séries de preços muito diferentes, como Bitcoin e Ethereum, uma delas costuma “achatar” a outra no gráfico, dificultando a análise visual. Para superar esse problema, utilizamos uma normalização baseada em índice, também chamada de base 100. Ela transforma cada série para que ambas comecem com o valor 100, e a partir disso passam a representar apenas o crescimento relativo ao longo do tempo. Isso é calculado da seguinte maneira:

$$
    \text{Valor Normalizado} = \frac{\text{Valor Atual}}{\text{Valor Inicial}} \cdot 100
$$

Dessa forma, temos que a técnica:

* Coloca BTC e ETH na mesma escala, sem distorção;

* Permite comparar quem cresceu mais ao invés de quem é mais caro;

* É o método mais utilizado em análises financeiras para comparar ativos diferentes;

* Mantém uma interpretação intuitiva: 100 = valor inicial, 120 = subiu 20% e 80 = caiu 20%.

Um valor qualquer no eixo $y$ significa o quanto em $\%$ a criptomoeda específica variou em relação ao marco inicial.

```python
fig = plt.figure(figsize=(12,8))

#Normalização em base 100.
btc_norm = btc / btc.iloc[0] * 100
eth_norm = eth / eth.iloc[0] * 100

#Plotagem.
plt.plot(dias, eth_norm, label='ETH (normalizado)', color='green', linestyle='-')
plt.plot(dias, btc_norm, label='BTC (normalizado)', color='blue', linestyle='-')

plt.title('Desempenho Relativo - BTC vs ETH (Base 100)')
plt.xlabel('Dias')
plt.ylabel('Desempenho Relativo(%)')

plt.legend()

# Ticks do eixo x.
indices = np.linspace(0, len(dias)-1, 6)
plt.xticks(dias.iloc[indices])

#Médias das séries normalizadas.
m_btc = btc_norm.mean()
m_eth = eth_norm.mean()

plt.axhline(y=m_btc, color='blue', linestyle='--', linewidth=1)
plt.axhline(y=m_eth, color='green', linestyle='--', linewidth=1)

# Ticks do eixo y com inserção das médias.
indices_y = np.linspace(0, max(btc_norm.max(), eth_norm.max()), 5)
plt.yticks(list(indices_y) + [m_btc, m_eth])

# Deixar as médias em vermelho no eixo Y.
valores, legendas = plt.yticks()
for val, leg in zip(valores, legendas):
    if val in [m_btc]:
        leg.set_color('blue')
    elif val in [m_eth]:
        leg.set_color('green')

plt.show()
```

##### **Z-Escore**

O Z-score (ou escore-z) é uma forma de padronizar dados, mostrando quantos desvios-padrão um valor está da média da série. Ou seja:

$$
    z = \frac{x_i - \mu}{\sigma}
$$

A Interpretação é a seguinte:

* $z = 0$, então o valor está na média;

* $z > 0$, então o valor está acima da média;

* $z < 0$, então valor está abaixo da média.

Quanto maior o módulo de $z$, mais fora do padrão o valor está. Além disso, o z-escore costuma ficar entre $\pm 2$, valores com módulo acima de $3$ indicam valores atípicos (*outliers*).

Podemos fazer o gráfico também de z-escore dos nossos dados:

```python
fig = plt.figure(figsize=(12, 8))

#Cálculo do escore-z.
btc_z = (btc - btc.mean())/btc.std()
eth_z = (eth - eth.mean())/eth.std()

plt.plot(dias, btc_z, label = "btc", color = "blue", linestyle = "-")
plt.plot(dias, eth_z, label = "eth", color = "green", linestyle = "-")

plt.title('Desempenho Relativo - BTC vs ETH (Z-score)')
plt.xlabel('Dias')
plt.ylabel('Z-score')

plt.legend()

# Ticks do eixo x
indices = np.linspace(0, len(dias)-1, 6, dtype=int)
plt.xticks(dias.iloc[indices])

# Linhas horizontais em Z = 0 (média).
plt.axhline(y=0, color='black', linestyle='--', linewidth=1)

plt.show()
```

Perceba que o z-escore premite detectar altos e baixos fora do comum, evidenciando picos, quedas ou anomalias, e facilita comparar séries diferentes, mostrando de forma padronizada quem está acima ou abaixo do esperado.

##### 5.2) **Analisando Diferenças com Histogramas**

Vamos construir visualizações das diferenças de preços entre cada dia com gráficos do formato **histograma**. Para isso, iremos utilizar a função `.diff()` para retornar as diferenças diárias.

```python
btc_dif = btc.diff()
btc_dif.head()
```

```python
eth_dif = eth.diff()
eth_dif.head()
```

Agora vamos fazer histogramas! Um histograma é uma representação gráfica da distribuição de frequência de um conjunto de dados quantitativos (quantas vezes algo aparece dado um intervalo). Ele organiza os valores em intervalos contínuos ou classes e utiliza barras para indicar a frequência de observações que pertencem a cada intervalo. Dessa forma, intervalos menores permitem uma análise mais detalhada da distribuição, embora intervalos excessivamente pequenos possam dificultar a interpretação. Veja a seguir os histogramas para as diferenças de preços das duas criptomoedas:

```python
fig = plt.figure(figsize=[12,8])

#Para histogramas (gráfico de frequências em intervalos), usamos hist.
plt.hist(eth_dif, label = "eth", color = "green")
plt.hist(btc_dif, label = "btc", color = "blue")

plt.title("Variação diária - Preços")
plt.xlabel("Variação ($USD)")
plt.ylabel("Frequência")

plt.legend()

plt.show()
```

Tudo bem, temos os dois histogramas! Mas temos um problema, os dois estão sobrepostos, e isso é natural. Para tentar resolver isso, vamos aumentar a transparência:

```python
fig = plt.figure(figsize=[12,8])

#Lembremos do alpha!
plt.hist(eth_dif, label = "eth", color = "green", alpha = 0.6)
plt.hist(btc_dif, label = "btc", color = "blue", alpha = 0.6)

plt.title("Variação diária - Preços")
plt.xlabel("Variação ($USD)")
plt.ylabel("Frequência")

plt.legend()

plt.show()
```

Vamos aumentar o número de intervalos (ou *bins*) nos histogramas:

```python
fig = plt.figure(figsize=[12,8])

#Use bins para aumentar os intervalos!
plt.hist(eth_dif, label = "eth", color = "green", alpha = 0.6, bins = 20)
plt.hist(btc_dif, label = "btc", color = "blue", alpha = 0.6, bins = 50)

plt.title("Variação diária - Preços")
plt.xlabel("Variação ($USD)")
plt.ylabel("Frequência")

plt.legend()

plt.show()
```

Vamos plotar linhas para as médias das diferenças:

```python
fig = plt.figure(figsize=[12,8])

#Argumento bins diz o número de intervalos contíguos.
plt.hist(eth_dif, label = "eth", color = "green", alpha = 0.6, bins = 20)
plt.hist(btc_dif, label = "btc", color = "blue", alpha = 0.6, bins = 50)

#Linhas verticais com axvline.
plt.axvline(btc_dif.mean(), color = "red", linestyle = "--", linewidth = 0.8)
plt.axvline(eth_dif.mean(), color = "red", linestyle = "--", linewidth = 0.8)

plt.title("Variação diária - Preços")
plt.xlabel("Variação ($USD)")
plt.ylabel("Frequência")

plt.legend()

plt.show()
```

Como as retas das médias estão sobrepostas, isso significa que a média da variação de ambas as moedas está sempre rondando próximo do valor zero. No entanto, o alcance de variação do preço do *BitCoin* mostra que sua escala é bem maior, engolindo o histograma da *Ethereum*. Vamos dessa vez calcular a variação percentual, ao invés de ver valores brutos! Para isso, ao invés de usar `.diff`, iremos utilizar `.pct_change`, que calcula a variação percentual (0 a 1) de um dado e seu anterior em um dataframe.

```python
#Usamos pct_change e multiplicamos por 100 para termos em %.
btc_pct = btc.pct_change() * 100  
eth_pct = eth.pct_change() * 100  

#Dropamos os valores iniciais sem valores (NaN), já que não possuem dado anterior (boa prática).
btc_pct = btc_pct.dropna()
eth_pct = eth_pct.dropna()

fig = plt.figure(figsize=(12, 8))

#Histogramas.
plt.hist(btc_pct, bins=30, label='BTC', color='blue', alpha=0.6, edgecolor='black', linewidth=0.5)
plt.hist(eth_pct, bins=30, label='ETH', color='green', alpha=0.7, edgecolor='black', linewidth=0.5)

# Linhas de média.
plt.axvline(btc_pct.mean(), color='darkblue', linestyle='--', linewidth=0.5, 
            label=f'Média BTC: {btc_pct.mean():.2f}%')
plt.axvline(eth_pct.mean(), color='darkgreen', linestyle='--', linewidth=0.5, 
            label=f'Média ETH: {eth_pct.mean():.2f}%')

plt.title(f'Distribuição de Variação Diária - BTC vs ETH')
plt.xlabel('Variação Diária (%)')
plt.ylabel('Frequência')
plt.legend()

plt.show()
```

Impressionante! As escalas diferentes estavam mascarando informações importantes:

* Embora a sua média de variação continuem rondando zero, esse novo gráfico demonstra que a Ethereum possui maior incidência de vezes próximos ao zero;

* A Ethereum possui maior incidência de anomalias, ou seja, valores "extremos" são mais comuns, o que ficava mascarado nos histogramas anteriores;

* Bitcoin definitivamente é a criptomoeda mais está estável em relação a sua escala!

O comportamento é então algo parecido como:

* BTC: aparenta certa previsibilidade, movimentos suaves, mas constantes.

* ETH: movimento suave rondando zero, mas com algumas altas e baixas mais agudas.

```python
fig, (ax_cima, ax_baixo) = plt.subplots(figsize = (12, 8), nrows = 2)
```
