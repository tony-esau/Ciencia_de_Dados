#### **O que é amostragem?**

O processo de amostragem é formalmente definido como a seleção de um subconjunto (denominado amostra) de membros de uma população ou conjunto de dados original, com o propósito de derivar estimativas e tirar conclusões sobre essa população utilizando apenas os membros selecionados. Matematicamente, se $N$ representa o número total de elementos na população e $n$ o tamanho da amostra $S$, o processo seleciona elementos de tal forma que $n<N$

#### **Característica Essencial de uma Amostragem:**

Para que a amostragem seja válida, a amostra deve ser **representativa**, o que significa que ela deve conservar as características e propriedades consideradas importantes do conjunto original. Assim, os resultados obtidos pela análise da amostra devem ser muito semelhantes aos que seriam encontrados caso toda a população fosse processada


#### **Importância**

A importância da amostragem reside na viabilização da análise de dados em larga escala através de diversos benefícios:

- **Redução do Volume e Custo Computacional**: A mineração de dados e a análise de grandes volumes são processos caros e complexos. A amostragem funciona como uma técnica de redução de dados que permite que algoritmos operem de forma mais eficiente e rápida, diminuindo a complexidade de tempo;

- **Economia de Recursos**: Em muitas aplicações, coletar dados de toda uma população é irrealista ou impossível devido aos altos custos financeiros, de tempo, materiais e de mão de obra. A amostragem permite obter informações confiáveis com gastos significativamente menores;

- **Agilidade e Tempestividade**: Por ser um processo mais rápido, a amostragem permite que os resultados sejam publicados e utilizados de forma oportuna. Em ciência de dados, uma resposta aproximada e rápida é frequentemente preferível a uma resposta exata que demore muito tempo para ser processada;

- **Melhoria na Qualidade dos Dados**: Como o volume de dados a ser tratado é menor, os investigadores podem ser mais cuidadosos na coleta, treinamento e supervisão, o que muitas vezes resulta em dados mais precisos do que os obtidos em um censo completo, que pode estar sujeito a erros administrativos e de coleta em larga escala;

- **Necessidade Técnica**: Em casos onde a medição exige a destruição da unidade de observação (como testes de laboratório), a amostragem é o único meio de obter informações sem eliminar toda a população.

 #### **Amostragem por Reservatório**

A Amostragem Polarizada por Densidade é uma técnica de redução de dados onde a probabilidade de um elemento ser selecionado para a amostra depende da densidade local de sua vizinhança. Diferente da amostragem uniforme, ela busca equilibrar a representatividade de agrupamentos (clusters) com tamanhos ou densidades muito distintos, realizando o *oversampling* (superamostragem) de regiões esparsas e o *undersampling* (subamostragem) de regiões densas.

Considere um conjunto de dados com $N$ pontos onde se deseja extrair uma amostra de tamanho $n$. O processo segue esta lógica geral:
- Estimação de Densidade: Para cada elemento $x$, calcula-se uma função de densidade $f(x)$ que quantifica a concentração de pontos em sua vizinhança local;
- Probabilidade de Inclusão: A probabilidade $P(x)$ de um ponto ser incluído na amostra é definida como: 
    - $P(x) = \frac{f^a(x)}{\sum_{i=1}^{N} f^a(x_i)}$, onde:
        - $\sum_{i=1}^{N} f^a(x)$ para garantir que a soma das probabilidades seja coerente;
        - $a$ é um parâmetro de ajuste. Quando $a<0$, o algoritmo privilegia pontos em regiões de baixa densidade (bias negativo), aumentando a chance de selecionar elementos de grupos raros. Se $a > 0$, favorece regiões de alta densidade.

O algoritmo mais simples, Density-Based Sampling (DBS), funciona da seguinte maneira 
1. **Cálculo da Densidade Local** ($f(x)$): Para cada ponto do conjunto de dados, você deve estimar a densidade de sua vizinhança. Isso é feito medindo a distância (geralmente a Euclidiana) entre os pontos. Pode-se usar:
    - $\text{Kernel Density Estimation}$ (KDE): Soma-se a influência de todos os pontos próximos usando uma função de kernel (como a Gaussiana);
    - $\text{K-vizinhos mais próximos}$: A densidade é estimada com base na distância média para os $k$ vizinhos mais próximos.

2. **Definição do Parâmetro de Ajuste** ($a$): Você define um expoente a que determina o "viés" da amostragem:
- Para priorizar regiões de baixa densidade (capturar grupos raros), use um $a$ negativo (ex: $−0,25$ ou $−0,5$), isso garante que pontos em áreas esparsas tenham uma probabilidade maior de seleção.

3. **Cálculo da Probabilidade de Inclusão** [$P(x) = \frac{f^a(x)}{\sum_{i=1}^{N} f^a(x_i)}$].

4. **Sorteio e Seleção**: Para cada ponto, gera-se um número aleatório. Se esse número for menor ou igual a $P(x)$, o ponto é adicionado à amostra final.

##### **Cálculo de tamanho**

Para esse exemplo, iremos calcular o tamanho da amostra a partir do Intervalo de Confiança para Proporção:

$$\hat{p} \pm Z \cdot \sqrt{\frac{\hat{p}(1 - \hat{p})}{n}}$$

Em que:

- $\hat{p}$: Proporção observada na amostra (estimativa da proporção real).

- $Z$: Valor crítico da distribuição normal padrão (depende do nível de confiança).

- $n$: Tamanho da amostra.

- $\hat{p}(1 - \hat{p})$: Variância da proporção.

- $\sqrt{\frac{\hat{p}(1 - \hat{p})}{n}}$: Erro padrão da proporção (mede a variabilidade da estimativa).

Vamos calcular considerando $99\%$ de confiança e erro menor igual a $2\%$. Para isso devemos garantir que o erro padrão de proporção seja menor ou igual a $2\%$. Para isso, consideremos o pior caso para o denominador $\hat{p}(1-\hat{p})$, isto é, quando ele é maximizado. Um estudo quadrático simples, nos apontará que esse valor é $0,5$. Vamos rearranjar a fórmula sabendo disso:

$$Z \cdot \sqrt{\frac{\hat{p}(1 - \hat{p})}{n}} \leq 0,02$$
$$ \frac{1}{Z}Z \cdot \sqrt{\frac{\hat{p}(1 - \hat{p})}{n}} \leq \frac{1}{Z} 0,02$$
$$\sqrt{\frac{\hat{p}(1 - \hat{p})}{n}} \leq \frac{0,02}{Z}$$
$$\frac{\sqrt{(0,5)^2}}{\sqrt{n}} \leq \frac{0,02}{Z}$$
$$\frac{0,5}{\sqrt{n}} \leq \frac{0,02}{Z}$$
$$\frac{\sqrt{n}} {0,5} \geq \frac{Z}{0,02}$$
$$ \sqrt{n} \geq \frac{Z}{0,02} \cdot 0,5$$
$$ n \geq (\frac{Z}{0,04})^2$$



Calculando:

```python
from scipy.stats import norm
import math

Z = norm.ppf(0.99)

n = math.ceil((Z/0.04)**2)
n
```

 #### **Amostragem Polarizada por Densidade (DBS)**


```python
# Tomando as colunas de dados gerados pelo PCA.
features = [col for col in dados.columns if col.startswith('V')]
X = dados[features]
X
```

Vamos usar os $k$ vizinhos mais próximos.

```python
from sklearn.neighbors import NearestNeighbors
import numpy as np

# O k pode ser ajustado dependendo do tamanho do dataset e da dimensionalidade, o interessante seria realizar testes para encontrar o melhor valor.
k = 10

# Calcula as distâncias para os k vizinhos mais próximos de cada ponto.
nbrs = NearestNeighbors(n_neighbors=k).fit(X)

distancias, _ = nbrs.kneighbors(X)

"""
    - Média das distâncias aos vizinhos.
    - A adição de um pequeno valor (1e-8) evita divisão por zero.
    - A densidade é inversamente proporcional à média das distâncias: pontos em regiões mais densas terão distâncias menores, resultando em pesos maiores.
"""

densidade = 1 / (distancias.mean(axis=1) + 1e-8)
```

```python
a = -100  # Penalizar as amostras em regiões de alta densidade.
"""
    - O valor de a pode ser ajustado para controlar o grau de polarização: valores mais negativos aumentam a penalização para regiões densas, enquanto
    valores próximos de zero reduzem essa penalização.
"""
peso = densidade ** a
```

```python
# Normaliza os pesos para obter probabilidades.
P = peso / peso.sum()
```

```python
# Amostra os índices dos dados com base nas probabilidades calculadas.
indices = np.random.choice(len(dados), size=n, replace=False, p=P)

# Seleciona as amostras correspondentes aos índices amostrados.
amostra = dados.iloc[indices]
amostra
```

Vamos observar a proporção:

```python
dados['Class'].value_counts(normalize=True)
```

```python
amostra['Class'].value_counts(normalize=True)
```

Usando $a=-100$ temos um valor razoável, mas perceba que o alto desbalanceamento dificulta. Vale ressaltar também que nesse método a penalização pode facilitar a entrada de *outliers* de todas as classes, pois não faz distinção entre elas, o método leva em consideração simplesmente a densidade e geometria.

Podemos aplicar essa ideia selecionando todos elementos da classe minoritária e aplicar o método na outra classe:

```python
fraudes = dados[dados['Class'] == 1]
```

```python
normais = dados[dados['Class'] == 0]
```

```python
features = [col for col in dados.columns if col.startswith('V')]
X_normais = normais[features]

# Calcula densidade.
nbrs = NearestNeighbors(n_neighbors=11).fit(X_normais)
dist, _ = nbrs.kneighbors(X_normais)
dist = dist[:, 1:]

dens = 1 / (dist.mean(axis=1) + 1e-8)
```

```python
# Aplica viés.
a = -1
peso = dens ** a
P = peso / peso.sum()

# Amostra.
n_normais = n - len(fraudes)
idx = np.random.choice(len(normais), size=n_normais, replace=False, p=P)

amostra_normais = normais.iloc[idx]
amostra_1 = pd.concat([fraudes, amostra_normais], ignore_index=True)
amostra_1
```

 #### **Amostragem Polarizada por Densidade (BBS)**


O algoritmo BBS (Biased Box Sampling), é uma técnica de mineração de dados que seleciona elementos de forma que a probabilidade de inclusão dependa da densidade local da vizinhança. O objetivo é reduzir o volume de dados preservando a representatividade de agrupamentos (clusters) desbalanceados, pequenos ou raros, que seriam ignorados por uma amostragem uniforme. O BBS utiliza um mapeamento de multi-resolução do espaço de dados através de uma estrutura de árvore ($\text{MG-tree}$):
- Divisão do Espaço: O espaço é dividido recursivamente em $2^E$ regiões (onde $E$ é o número de atributos), dividindo cada atributo ao meio;
- Limiar de Divisão ($\delta$): Uma região só continua sendo dividida se o número de pontos for superior a $\delta$, definido como:
    - $\delta=\frac{2 \cdot E×100}{\text{Ratio}}​$, em que $\text{Ratio}$ é a taxa de amostragem desejada.
- Cálculo do Tamanho da Amostra (*samplesize*): Para cada nó folha na árvore compacta (MGc-tree), o número de pontos selecionados é ajustado pela sua profundidade para realizar um *oversampling* em regiões menos densas:
$$n = \left( \frac{Ratio \times C_{r,i}}{100} \right) \times 2 \times (R - \text{Nivel\_Atual})$$
em que:

- $C_{r,i}$​ é a contagem de pontos na célula, $R$ é o nível máximo e $\text{Nivel\_Atual}$ é o nível atual do nó. O custo total é linear: $O(N⋅E)$.

```python
features = [col for col in dados.columns if col.startswith('V')]

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(dados[features])
X_scaled = pd.DataFrame(X_scaled, columns=features)
```

```python
grid_size = 5
X_disc = (X_scaled * grid_size).astype(int)
```

```python
celulas = X_disc.groupby(list(X_disc.columns)).size()
```

```python
densidade = X_disc.merge(
    celulas.rename("count"),
    left_on=list(X_disc.columns),
    right_index=True
)["count"]
```

```python
peso = 1 / densidade
```

```python
P = peso / peso.sum()

# Gera mais pontos do que precisa.
intermediate_size = int(len(dados) * 0.2)  # 20% (ajustável).

idx_inter = np.random.choice(len(dados),
                             size=intermediate_size,
                             replace=False,
                             p=P)

amostra_inter = dados.iloc[idx_inter]
```

```python
final_size = 3383

idx_final = np.random.choice(len(amostra_inter),
                            size=final_size,
                            replace=False)

amostra_bbs = amostra_inter.iloc[idx_final]
amostra_bbs
```
